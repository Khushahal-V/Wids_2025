```python
from transformers import pipeline
import textwrap

summarizer = pipeline("summarization")

text = (
    "In late 2025, the conflict between Russia and Ukraine sharply intensified, "
    "with Russia launching large aerial attacks on Kyiv that damaged critical infrastructure "
    "and left many without power, complicating peace efforts and international talks. "
    "At the same time, China is rapidly expanding its nuclear weapons manufacturing capacity, "
    "raising global security concerns about regional arms races and accidental escalation. "
    "In the Middle East, Israel has carried out airstrikes in southern Lebanon amid unresolved tensions "
    "with Hezbollah as ceasefire deadlines near, further destabilizing the region. "
    "Meanwhile, Myanmar’s controversial elections have taken place under heavy influence from China, "
    "drawing criticism that real political power remains outside the hands of ordinary citizens. "
    "In West Africa’s Sahel region, Mali, Burkina Faso, and Niger have launched joint operations "
    "against extremist groups to address persistent insecurity and bolster regional cooperation."
)

summary = summarizer(
    text,
    min_length=40,
    max_length=100,
    do_sample=False
)

original_length = len(text.split())
summary_text = summary[0]["summary_text"]
summary_length = len(summary_text.split())

print("Original Text:")
print(textwrap.fill(text, width=80))
print("\nOriginal Length (words):", original_length)

print("\nGenerated Summary:")
print(textwrap.fill(summary_text, width=80))
print("\nSummary Length (words):", summary_length)
```

```python
from transformers import pipeline

generator = pipeline("text-generation")

prompt = "In 2030, AI systems will"

outputs = generator(
    prompt,
    max_new_tokens=50,
    num_return_sequences=2,
    do_sample=True
)

print("Prompt:")
print(textwrap.fill(prompt, width=80))

print("\nGenerated Text 1:")
print(textwrap.fill(outputs[0]["generated_text"], width=80))

print("\nGenerated Text 2:")
print(textwrap.fill(outputs[1]["generated_text"], width=80))
```

```python
analyzer = pipeline("sentiment-analysis")

reviews = [
    "The movie was absolutely fantastic and kept me engaged throughout.",
    "I found the film boring and a complete waste of time.",
    "The acting was decent but the story was predictable.",
    "An amazing cinematic experience with stunning visuals.",
    "The plot was weak and the characters were poorly developed."
]

results = analyzer(reviews)

for review, result in zip(reviews, results):
    print("Review:")
    print(review)
    print("Sentiment:", result["label"])
    print("Confidence Score:", round(result["score"], 4))
    print("-" * 50)
```
